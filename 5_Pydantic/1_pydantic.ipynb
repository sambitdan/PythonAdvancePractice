#### Pydantic is a python library for data validation and parsing

```python
from pydantic import BaseModel # BaseModel is class
```

```python
class employee_shema(BaseModel):
    name : str
    id : int
    department : str

# After inheriting BaseModel our employee_schema class has become Pydantic object
```

```python
obj = employee_shema(name = 'John Doe', id = 123, department = "HR") #Pydantic Object
obj
```

```python
print(obj)
```

```python
obj2 = employee_shema(**{"name" : 'John Doe', "id" : 123, "department" : "HR"})
```

```python
obj2
```

```python
obj2 = employee_shema(**{"name" : 'John Doe', "id" : 123, "department" : 101})
obj2
```

### Coercion vs StrictTypes

### Coercion

```python
obj2 = employee_shema(**{"name" : 'John Doe', "id" : "123", "department" : "HR"})
obj2 # This is called coercion or linient behaviour where pydantic will try to convert our datatypes into defined datatypes instead of throwing errors
```

```python
obj2 = employee_shema(**{"name" : 'John Doe', "id" : "ABC", "department" : "HR"})
obj2 # This is called coercion or linient behaviour where pydantic will try to convert our datatypes into defined datatypes instead of throwing errors
```

### StrictTypes

```python
from pydantic import StrictInt
```

```python
class employee_shema(BaseModel):
    name : str
    id : StrictInt
    department : str


obj4 = employee_shema(**{"name" : 'John Doe', "id" : "123", "department" : "HR"})
obj4  # As this is strict type it will not convert integer
```
